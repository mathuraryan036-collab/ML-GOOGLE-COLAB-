Imbalanced Dataset - A dataset with an unequal class distribution

```python
# importing the dependencies
import numpy as np
import pandas as pd
```

```python
# loading the dataset to the pandas Dataframe
credit_card_data = pd.read_csv('/content/credit_data.csv')
```

```python
# first 5 rows of the dataframe
credit_card_data.head()
```

```python
credit_card_data.tail()
```

```python
# distribution of the Two classes
credit_card_data['Class'].value_counts()
```

This is highly imbalanced dataset

0 --> Legit Transactions

1 --> Fraud Transactions

```python
# separating the legit and fraud transactions
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

Undersampling

Build a sample dataset containing similar distribution of Legit and Fraud Transactions

Number of Fraud Transactions --> 371

```python
legit_sample = legit.sample(n=371)
```

```python
print(legit_sample.shape)
```

Concatenate the Two Dataframes

```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```
