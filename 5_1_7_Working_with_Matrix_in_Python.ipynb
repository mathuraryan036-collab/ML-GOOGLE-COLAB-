```python
import numpy as np
```

Creating a Matrix using Numpy

```python
matrix_1 = np.array([[2,3],[6,7]])
print(matrix_1)
```

```python
matrix_1.shape
```

```python
matrix_2 = np.array([[5,6,7],[6,7,9],[4,5,6]])
print(matrix_2)
```

```python
matrix_2.shape
```

Creating Matrices with Random Values

```python
random_matrix = np.random.rand(3,3)
print(random_matrix)
```

```python
# creating matrix with random integer
random_integer_matrix = np.random.randint(100, size=(4,5))
print(random_integer_matrix)
```

Matrix with all the values as 1

```python
matrix_3 = np.ones((3,4))
print(matrix_3)
```

```python
matrix_3 = np.ones((3,4), dtype=int)
print(matrix_3)
```

Null Matrix

```python
null_matrix = np.zeros((3,4))
print(null_matrix)
```

```python
null_matrix = np.zeros((3,4), dtype=int)
print(null_matrix)
```

Identity Matrix

```python
identity_matrix = np.eye(5)
print(identity_matrix)
```

```python
identity_matrix = np.eye((5), dtype = int)
print(identity_matrix)
```

Transpose of a Matrix

```python
# matrix with random integer values
a = np.random.randint(100, size=(4,5))
print(a)
```

```python
transpose_of_a = np.transpose(a)
print(transpose_of_a)
```
